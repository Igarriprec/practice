```python
temperature=-3
```

```python
temperature = -3

if temperature > 35:
    print("Severe heat warning! Stay indoors.")
elif 21 <= temperature <= 35:
    print("Pleasant weather.")
elif temperature == 16:
    print("Baseline AC calibration temperature reached.")
elif 0 <= temperature < 21 and temperature != 16:
    print("Chilly day. Grab a jacket.")
else:
    print("Freezing alert! Watch for ice.")
```

```python
employees = [
    {"id": 1, "name": "Amina", "department": "Engineering", "hours": 42},
    {"id": 2, "name": "Chidi", "department": "Marketing", "hours": 38},
    {"id": 3, "name": "Elena", "department": "Engineering", "hours": 45},
    {"id": 4, "name": "Tariq", "department": "Support", "hours": 35},
    {"id": 5, "name": "Sarah", "department": "Engineering", "hours": 40}
]
```

```python
for employee in employees:
    if employee["hours"] > 40:
        print(f"[Overtime Alert] {employee['name']} worked {employee['hours']} hours.")
    else:
        print(f"[Standard] {employee['name']} worked {employee['hours']} hours.")
```
